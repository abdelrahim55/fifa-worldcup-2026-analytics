```python
player_df = pd.read_csv("FIFA_Data/data/raw/player_stats.csv")

player_df.info()
player_df.head()
player_df.isnull().sum()
player_df.duplicated().sum()
```

```python
os.makedirs("processed", exist_ok=True)
```

```python
matches_df = pd.read_csv(r"FIFA_Data/data/raw/matches.csv")
matches_df.head()
matches_df.columns
matches_df.info()
```

```python
matches_df.isnull().sum()
```

```python
matches_df["home_penalty_score"] = (
    matches_df["home_penalty_score"]
    .fillna(0)
    .astype(int)
)

matches_df["away_penalty_score"] = (
    matches_df["away_penalty_score"]
    .fillna(0)
    .astype(int)
)
```

```python
matches_df.isnull().sum()
```

```python
matches_df["date"] = pd.to_datetime(matches_df["date"])
```

```python
matches_df.info()
```

```python
os.makedirs("FIFA_Data/processed", exist_ok=True)

matches_df.to_csv(
    "processed/matches_clean.csv",
    index=False
)
os.listdir("processed")
```

```python
squads_and_players = pd.read_csv(r"FIFA_Data/data/raw/squads_and_players.csv")
squads_and_players.info()
```

```python
squads_and_players.isnull().sum()
```

```python
squads_and_players.duplicated().sum()
```

```python
squads_and_players["date_of_birth"] = pd.to_datetime(
    squads_and_players["date_of_birth"]
)
```

```python
text_cols = ["player_name", "position", "club_team"]

for col in text_cols:
    squads_and_players[col] = squads_and_players[col].str.strip()
```

```python
numeric_cols = [
    "market_value_eur",
    "caps",
    "height_cm",
    "goals"
]

(squads_and_players[numeric_cols] < 0).sum()
```

```python
squads_and_players.to_csv(
    "processed/squads_and_players_clean.csv",
    index=False
)
os.listdir("processed")
```
